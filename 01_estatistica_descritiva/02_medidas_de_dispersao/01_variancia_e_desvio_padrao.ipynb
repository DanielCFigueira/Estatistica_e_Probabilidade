# Implementação prática da Variância e do Desvio Padrão em Python.

* Aplicações da Variância e do Desvio Padrão dentro do universo de IA e Dados:

* Códigos exemplares:

```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import StandardScaler

df = pd.read_csv("../../data/dataset_estatistica.csv")

# 1. Cálculo da Variância e Desvio Padrão:
variancia_renda = df["renda_mensal"].var()
desvio_padrao_renda = df["renda_mensal"].std()

print(f"Renda Mensal Média: R$ {df['renda_mensal'].mean():.2f}")
print(f"Variância: {variancia_renda:.2f}")
print(f"Desvio Padrão: R$ {desvio_padrao_renda:.2f}")

# 2. Exemplo de Padronização (Z-Score) em uma variável sem nulos:
df_limpo = df.dropna(subset=['renda_mensal']).copy()

scaler = StandardScaler()

df_limpo["renda_zscore"] = scaler.fit_transform(df_limpo[["renda_mensal"]])

print("\nDados com Z-Score aplicado (Média ~ 0, Desvio Padrão ~ 1):")
display(df_limpo[["renda_mensal", "renda_zscore"]].head())
```
