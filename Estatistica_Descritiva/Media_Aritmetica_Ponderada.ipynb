# Implementação prática em Python.

```python
import numpy as np
import pandas as pd

# 1. Criação de um DataFrame simulado de logs de transações:

data = {
    'request_id': range(1,6),
    'latency_ms': [12.5, 15.0, 11.2, 340.8, 14.1],
    'precision_weight': [1.0, 1.0, 0.9, 0.2, 1.0]
}
df = pd.DataFrame(data)

# --- MÉDIA ARITMÉTICA SIMPLES ---

media_simples_pandas = df['latency_ms'].mean()
media_simples_numpy = np.mean(df['latency_ms'].to_numpy())

print(f"Média Aritmética Simples (Pandas): {media_simples_pandas:.2f} ms")
print(f"Média Aritmética Simples (Numpy): {media_simples_numpy:.2f} ms")

# --- MÉDIA PONDERADA ---
# Calculada combinando a multiplicação vetorizada de arrays com a soma das colunas
# Formula: sum(value * weight) / sum(weight)

valores = df['latency_ms']
pesos = df['precision_weight']
media_ponderada = np.average(valores, weights=pesos)
print(f"Média Ponderada: {media_ponderada:.2f} ms")
```
