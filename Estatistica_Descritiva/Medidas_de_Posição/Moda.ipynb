# Aplicações práticas do uso da Moda em Python.

* A Moda é a medida de posição e tendência central definida como a realização mais frequente ou de maior ocorrência em um conjunto de dados. Diferentemente da Média e da Mediana — que exigem operações de soma ou ordenação de valores —, a moda fundamenta-se puramente na contagem de frequências. Por essa razão, ela é a única medida de localização estatística que pode ser aplicada a variáveis qualitativas nominais (categorias sem ordem intrínseca).

* Exemplos práticos:

1. Imputação de Valores Ausentes em Variáveis Categóricas:

```python
import pandas as pd

df = pd.read_csv("../../data/dataset_estatistica.csv")

# Identificando a moda(categoria mais frequente):
categoria_mais_comum = df["categoria_frequente"].mode()[0]
print(f"Categoria mais frequente na moda: {categoria_mais_comum}")

# Imputando a moda em valores nulos de uma coluna categórica:
df["categoria_frequente"] = df["categoria_frequente"].fillna(categoria_mais_comum)
```

2. Implementando uma Previsão por Votação Majoritária (Ensemble Voting):

```python
import numpy as np
from scipy import stats

# Simulando previsões de 5 modelos de ML para 3 novos clientes
# Classes: 0 = Não comprou, 1 = Comprou

previsoes_modelos = np.array([
    [1, 1, 0, 1, 1], # Cliente 1 -> Maioria previu 1 
    [0, 0, 1, 0, 0], # Cliente 2 -> Maioria previu 0
    [1, 0, 1, 0, 1] # Cliente 3 -> Maioria previu 1
])

# Calculando a moda ao longo do eixo dos modelos(axis=1):
previsao_final, _ = stats.mode(previsoes_modelos, axis=1)

print("Previsão final por maioria de votos(Moda):")
print(previsao_final.ravel())
```
