# Aplicações práticas da Média Truncada em Python.

* A **Média Truncada** (também chamada de **Média Aparada** ou *Trimmed Mean*) é uma estatística de localização concebida para combinar a eficiência da média aritmética com a resistência (robustez) da mediana. Ela resolve diretamente a principal vulnerabilidade da média simples: a extrema sensibilidade a valores discrepantes(outliers).

* Aplicações práticas da Média Truncada dentro de IA e Dados:

* Exemplo no Dataset do Projeto:

```python
import numpy as np
import pandas as pd
from scipy import stats

df = pd.read_csv("../../data/dataset_estatistica.csv")
renda_valida = df["renda_mensal"].dropna()

# 1. Média Simples (sofre impacto direto dos outliers):
media_simples = renda_valida.mean()

# 2. Mediana (descarta a forma da distribuição interna):
mediana = renda_valida.median()

# 3. Média Truncada em 10% (corta 10% do topo e 10% da base):
# O parâmetro proportiontocut varia de 0.0 a 0.5

media_truncada_10 = stats.trim_mean(renda_valida, proportiontocut=0.10)

print(f"Média Aritmética Simples : R$ {media_simples:.2f}")
print(f"Média Truncada (10%)    : R$ {media_truncada_10:.2f}")
print(f"Mediana                 : R$ {mediana:.2f}")
```
